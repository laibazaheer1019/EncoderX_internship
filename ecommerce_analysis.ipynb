```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# Read the excel file
df = pd.read_excel(
    "Data/Online Retail.xlsx",
    engine="openpyxl"
)

print("File loaded successfully!")
# ===============================================================================
# Step 2: Data Cleaning
# ==============================================================================
# Display the  total dimension of data
print("Dataset Shape:", df.shape)
print(df.head())
print("\nColumn Names:")
print(df.columns.tolist())
print("\nDataset Information:")
df.info()
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"], errors="coerce")
print("Missing InvoiceDate:", df["InvoiceDate"].isnull().sum())
# Check Duplicates
df = df.dropna(subset=["InvoiceDate"])
print("\nDuplicate Records:")
print(df.duplicated().sum())
df = df.drop_duplicates().copy()
print(df.shape)
# Find missing values
missing = df.isnull().sum()
missing_percentage = (missing / len(df)) * 100

missing_table = pd.DataFrame({
    "Missing Values": missing,
    "Percentage": missing_percentage
})

display(missing_table)

df = df.dropna(subset=["Description"])
print("Final shape after basic cleaning:", df.shape)
# Keep only positive quantities and prices
print("Negative Quantity:", (df["Quantity"] < 0).sum())
print("Zero Quantity:", (df["Quantity"] == 0).sum())
print("Negative UnitPrice:", (df["UnitPrice"] < 0).sum())
print("Zero UnitPrice:", (df["UnitPrice"] == 0).sum())
# cancelled  transactions
cancelled = df[df["InvoiceNo"].astype(str).str.startswith("C")]
print("Cancelled transactions:", len(cancelled))
print("Negative quantities:", (cancelled["Quantity"] < 0).sum())

negative_qty = df[df["Quantity"] < 0]

print(negative_qty[
    ["InvoiceNo", "StockCode", "Description", "Quantity", "UnitPrice"]
].head(10))

zero_price = df[df["UnitPrice"] == 0]

print(zero_price[
    ["InvoiceNo", "StockCode", "Description", "Quantity", "UnitPrice"]
].head(10))

negative_price = df[df["UnitPrice"] < 0]

print(negative_price[
    ["InvoiceNo", "StockCode", "Description", "Quantity", "UnitPrice"]
])
df = df[df["UnitPrice"] >= 0]

print("Shape after removing negative prices:", df.shape)
df_sales = df[df["Quantity"] > 0].copy()

print("Sales dataset shape:", df_sales.shape)
# Calculate Revenue first
df_sales["Revenue"] = df_sales["Quantity"] * df_sales["UnitPrice"]

Q1 = df_sales[["Quantity", "UnitPrice"]].quantile(0.25)
Q3 = df_sales[["Quantity", "UnitPrice"]].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
# Count outliers
for column in ["Quantity", "UnitPrice"]:
    outliers = (
        (df_sales[column] < lower_bound[column]) |
        (df_sales[column] > upper_bound[column])
    )

    print(column, "outliers:", outliers.sum())
# Cap (winsorize) the outliers
for column in ["Quantity", "UnitPrice"]:
    df_sales[column] = df_sales[column].clip(
        lower=lower_bound[column],
        upper=upper_bound[column]
    )
# Recalculate Revenue
df_sales["Revenue"] = df_sales["Quantity"] * df_sales["UnitPrice"]

print("\nOutliers treated using IQR capping.")
print(df_sales[["Quantity", "UnitPrice", "Revenue"]].describe())
# Save Clean Dataset
df_sales.to_csv("cleaned_online_retail.csv", index=False)
print("\nCleaned dataset saved successfully!")
# =================================================================================
# STEP 3: EXPLORATORY DATA ANALYSIS
# ================================================================================

# Basic Sales Summary
print("========== BASIC SALES ANALYSIS ==========")
print("Total Sales Records:", len(df_sales))

print("Total Revenue: {:.2f}".format(
    df_sales["Revenue"].sum()
))

print("Average Revenue per Record: {:.2f}".format(
    df_sales["Revenue"].mean()
))

print("Total Quantity Sold:",
      df_sales["Quantity"].sum())

print("Average Quantity per Record: {:.2f}".format(
    df_sales["Quantity"].mean()
))

print("Average Unit Price: {:.2f}".format(
    df_sales["UnitPrice"].mean()
))

print("Number of Unique Products:",
      df_sales["StockCode"].nunique())

print("Number of Unique Countries:",
      df_sales["Country"].nunique())

print("Number of Unique Customers:",
      df_sales["CustomerID"].nunique())

# Monthly Sales Analysis
df_sales["YearMonth"] = df_sales["InvoiceDate"].dt.to_period("M")

monthly_sales = (
    df_sales.groupby("YearMonth")["Revenue"]
    .sum()
    .reset_index()
)

print("Monthly Sales:")
display(monthly_sales)

highest_month = monthly_sales.loc[
    monthly_sales["Revenue"].idxmax()
]

lowest_month = monthly_sales.loc[
    monthly_sales["Revenue"].idxmin()
]

print("\nHighest Sales Month:")
print(highest_month)

print("\nLowest Sales Month:")
print(lowest_month)
# Top Products Analysis
top_products = (
    df_sales.groupby("Description")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("Top 10 Products by Revenue:")
display(top_products)

top_products_quantity = (
    df_sales.groupby("Description")["Quantity"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("\nTop 10 Products by Quantity Sold:")
display(top_products_quantity)
# Country Sales Analysis
country_sales = (
    df_sales.groupby("Country")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

print("Top 10 Countries by Revenue:")
display(country_sales.head(10))

highest_country = country_sales.idxmax()

print("\nHighest Revenue Country:", highest_country)
print("Revenue: {:.2f}".format(country_sales.max()))
# Customer Analysis
customer_sales = (
    df_sales.groupby("CustomerID")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

print("Top 10 Customers by Revenue:")
display(customer_sales.head(10))

print("\nAverage Revenue per Customer: {:.2f}".format(customer_sales.mean()))

# Correlation Analysis
correlation = df_sales[["Quantity", "UnitPrice", "Revenue"]].corr()

print(correlation.to_string())

print("\nQuantity vs Revenue:",
      correlation.loc["Quantity", "Revenue"])

print("UnitPrice vs Revenue:",
      correlation.loc["UnitPrice", "Revenue"])

print("Quantity vs UnitPrice:",
      correlation.loc["Quantity", "UnitPrice"])
# Anomaly Analysis
top_revenue = (
    df_sales[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "UnitPrice", "Revenue", "Country"]
    ]
    .sort_values("Revenue", ascending=False)
    .head(10)
)

print("Top 10 Highest Revenue Transactions:")
print(top_revenue.to_string(index=False))


top_quantity = (
    df_sales[
        ["InvoiceNo", "StockCode", "Description",
         "Quantity", "UnitPrice", "Revenue"]
    ]
    .sort_values("Quantity", ascending=False)
    .head(10)
)

print("\nTop 10 Highest Quantity Transactions:")
print(top_quantity.to_string(index=False))

# Exploratory Analysis Summary
print("========== EXPLORATORY ANALYSIS SUMMARY ==========")

print("Total Revenue:", round(df_sales["Revenue"].sum(), 2))
print("Total Quantity Sold:", df_sales["Quantity"].sum())
print("Number of Products:", df_sales["StockCode"].nunique())
print("Number of Countries:", df_sales["Country"].nunique())
print("Number of Customers:", df_sales["CustomerID"].nunique())

print("\nHighest Revenue Country:", highest_country)
print("Highest Revenue Product:", top_products.index[0])
print("Highest Revenue Month:", highest_month["YearMonth"])

print("\nQuantity-Revenue Correlation:",
      round(correlation.loc["Quantity", "Revenue"], 3))

print("UnitPrice-Revenue Correlation:",
      round(correlation.loc["UnitPrice", "Revenue"], 3))

print("Quantity-UnitPrice Correlation:",
      round(correlation.loc["Quantity", "UnitPrice"], 3))

# SAVE STEP 3 ANALYSIS RESULTS


# Save monthly sales analysis
monthly_sales.to_csv("monthly_sales_analysis.csv", index=False)

# Save top products by revenue
top_products.to_csv("top_products_by_revenue.csv")

# Save top products by quantity
top_products_quantity.to_csv("top_products_by_quantity.csv")

# Save country sales
country_sales.to_csv("country_sales_analysis.csv")

# Save top customers
customer_sales.to_csv("top_customers_by_revenue.csv")

# Save correlation matrix
correlation.to_csv("correlation_matrix.csv")

print("Step 3 analysis results saved successfully!")
# ==================================================================================
# STEP 4: DATA VISUALIZATION
# ==================================================================================
 # Bar Chart - Top 10 Products by Revenue

top_10_products = (
    df_sales.groupby("Description")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(12, 6))

top_10_products.sort_values().plot(kind="barh")

plt.title("Top 10 Products by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Product")
plt.tight_layout()
plt.savefig("top_10_products_revenue.png", dpi=300, bbox_inches="tight")
plt.show()

# Line Chart - Monthly Revenue Trend

plt.figure(figsize=(12, 6))

plt.plot(
    monthly_sales["YearMonth"].astype(str),
    monthly_sales["Revenue"],
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig("monthly_revenue_trend.png", dpi=300, bbox_inches="tight")
plt.show()

# Histogram - Revenue Distribution
plt.figure(figsize=(10, 6))

plt.hist(
    df_sales["Revenue"],
    bins=30
)

plt.title("Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Frequency")

plt.tight_layout()
plt.savefig("revenue_distribution.png", dpi=300, bbox_inches="tight")
plt.show()

# Correlation Heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.savefig("correlation_heatmap.png", dpi=300, bbox_inches="tight")
plt.show()

# Scatter Plot - Quantity vs Revenue
plt.figure(figsize=(10, 6))

plt.scatter(
    df_sales["Quantity"],
    df_sales["Revenue"],
    alpha=0.3
)

plt.title("Quantity vs Revenue")
plt.xlabel("Quantity")
plt.ylabel("Revenue")

plt.tight_layout()
plt.savefig("quantity_vs_revenue.png", dpi=300, bbox_inches="tight")
plt.show()
print("STEP 5: INSIGHTS AND RECOMMENDATIONS")
print()

print("KEY INSIGHTS")
print("1. United Kingdom was the dominant market, generating the highest revenue.")
print("2. November 2011 recorded the highest monthly revenue.")
print("3. REGENCY CAKESTAND 3 TIER was the highest-revenue product.")
print("4. JUMBO BAG RED RETROSPOT was the highest-selling product by quantity.")
print("5. Quantity had the strongest positive relationship with Revenue, with a correlation of 0.523.")
print("6. Revenue was concentrated among a small number of high-value customers.")
print("7. Quantity and UnitPrice showed a moderate negative correlation of -0.376.")
print()

print("RECOMMENDATIONS")
print("1. Focus marketing and customer-retention strategies on the United Kingdom market.")
print("2. Promote high-revenue products such as REGENCY CAKESTAND 3 TIER.")
print("3. Prepare inventory and marketing campaigns before peak sales periods such as November.")
print("4. Develop loyalty programs and personalized offers for high-value customers.")
print("5. Use both quantity and unit price when making product pricing and promotion decisions.")

```
